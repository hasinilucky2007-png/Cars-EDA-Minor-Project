```python
# Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset

df = pd.read_csv('/content/Car details v3.csv')

# Create a cleaned copy

df_clean = df.drop_duplicates().copy()

# Convert mileage

df_clean['mileage'] = pd.to_numeric(
    df_clean['mileage'].astype('string').str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Convert engine

df_clean['engine'] = pd.to_numeric(
    df_clean['engine'].astype('string').str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Convert max power

df_clean['max_power'] = pd.to_numeric(
    df_clean['max_power'].astype('string').str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Handle missing numerical values

numerical_columns = ['mileage', 'engine', 'max_power', 'seats']

for col in numerical_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Handle missing torque values

df_clean['torque'] = df_clean['torque'].fillna('Unknown')

# Display final information

print("Original Dataset:", df.shape)
print("Cleaned Dataset:", df_clean.shape)
print("Missing Values:", df_clean.isnull().sum().sum())
print("Duplicates:", df_clean.duplicated().sum())
```

```python
# ============================================================
# CARS DATASET - COMPLETE EDA MINOR PROJECT
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings("ignore")

from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Image,
    PageBreak, Table, TableStyle
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch
from reportlab.lib.enums import TA_CENTER

# Create output folder
os.makedirs("Cars_EDA_Output", exist_ok=True)

# ============================================================
# 1. DATASET OVERVIEW
# ============================================================

print("\n========== DATASET OVERVIEW ==========")

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
print("Total Duplicates:", df_clean.duplicated().sum())
print("Total Missing Values:", df_clean.isnull().sum().sum())

print("\nColumn Information:")
print(df_clean.info())

# ============================================================
# 2. OUTLIER ANALYSIS USING IQR
# ============================================================

print("\n========== OUTLIER ANALYSIS ==========")

numeric_cols = [
    'selling_price', 'km_driven',
    'engine', 'max_power'
]

outlier_results = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df_clean[col] < lower) |
             (df_clean[col] > upper)).sum()

    outlier_results.append([col, lower, upper, count])

    print(f"{col}: {count} potential outliers")

outlier_df = pd.DataFrame(
    outlier_results,
    columns=['Column', 'Lower Limit', 'Upper Limit', 'Outlier Count']
)

print("\n", outlier_df.to_string(index=False))

# Retain potential outliers; do not automatically delete them.

# ============================================================
# 3. DESCRIPTIVE STATISTICS
# ============================================================

print("\n========== DESCRIPTIVE STATISTICS ==========")

print(df_clean.describe().round(2))

print("\nCategorical Statistics:")
print(df_clean.describe(include='object'))

# ============================================================
# 4. CREATE FINAL CHARTS
# ============================================================

print("\n========== GENERATING CHARTS ==========")

chart_paths = []

def save_chart(filename):
    path = "Cars_EDA_Output/" + filename
    plt.tight_layout()
    plt.savefig(path, dpi=150, bbox_inches='tight')
    chart_paths.append(path)
    plt.show()
    plt.close()

# Chart 1: Top 10 Brands
df_clean['brand'] = df_clean['name'].str.split().str[0]

plt.figure(figsize=(9, 5))
df_clean['brand'].value_counts().head(10).plot(
    kind='barh'
)
plt.title("Top 10 Car Brands")
plt.xlabel("Number of Cars")
plt.ylabel("Car Brand")
plt.gca().invert_yaxis()
save_chart("01_top_brands.png")

# Chart 2: Fuel Type
plt.figure(figsize=(8, 5))
sns.countplot(data=df_clean, x='fuel')
plt.title("Distribution of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
save_chart("02_fuel_type.png")

# Chart 3: Selling Price Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df_clean['selling_price'], bins=40, kde=True)
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price (INR)")
plt.ylabel("Frequency")
save_chart("03_price_distribution.png")

# Chart 4: Year vs Selling Price
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_clean,
    x='year',
    y='selling_price',
    alpha=0.5
)
plt.title("Manufacturing Year vs Selling Price")
plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price (INR)")
save_chart("04_year_price.png")

# Chart 5: Kilometres vs Selling Price
sample = df_clean.sample(
    min(2000, len(df_clean)),
    random_state=42
)

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=sample,
    x='km_driven',
    y='selling_price',
    alpha=0.5
)
plt.title("Kilometres Driven vs Selling Price")
plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price (INR)")
save_chart("05_km_price.png")

# Chart 6: Transmission vs Average Price
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df_clean,
    x='transmission',
    y='selling_price',
    errorbar=None
)
plt.title("Average Selling Price by Transmission")
plt.xlabel("Transmission Type")
plt.ylabel("Average Selling Price (INR)")
save_chart("06_transmission.png")

# Chart 7: Correlation Heatmap
# Mileage excluded because original data has mixed units:
# kmpl and km/kg.

corr_cols = [
    'year', 'selling_price', 'km_driven',
    'engine', 'max_power', 'seats'
]

plt.figure(figsize=(8, 6))
sns.heatmap(
    df_clean[corr_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)
plt.title("Correlation Heatmap of Numerical Features")
save_chart("07_correlation.png")

# Chart 8: Outlier Boxplots
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_clean[numeric_cols])
plt.title("Outlier Detection Using Boxplots")
plt.xticks(rotation=20)
save_chart("08_outliers.png")

# ============================================================
# 5. KEY FINDINGS
# ============================================================

print("\n========== KEY FINDINGS ==========")

brand = df_clean['brand'].value_counts().idxmax()
fuel = df_clean['fuel'].value_counts().idxmax()
transmission = df_clean['transmission'].value_counts().idxmax()

avg_price = df_clean['selling_price'].mean()
avg_km = df_clean['km_driven'].mean()

transmission_price = df_clean.groupby(
    'transmission'
)['selling_price'].mean()

correlations = df_clean[corr_cols].corr()['selling_price'].drop(
    'selling_price'
)

strongest_feature = correlations.abs().idxmax()

findings = [
    f"The most common car brand is {brand}.",
    f"The most common fuel type is {fuel}.",
    f"The most common transmission type is {transmission}.",
    f"The average car selling price is Rs. {avg_price:,.2f}.",
    f"The average kilometres driven is {avg_km:,.2f}.",
    f"{strongest_feature} has the strongest correlation with selling price "
    f"among the analysed numerical features ({correlations[strongest_feature]:.2f})."
]

for i, finding in enumerate(findings, 1):
    print(f"{i}. {finding}")

print("\nAverage Price by Transmission:")
print(transmission_price.round(2))

# ============================================================
# 6. LIMITATIONS
# ============================================================

limitations = [
    "The dataset contains used-car listings and may not represent the entire car market.",
    "Some records originally contained missing or invalid values.",
    "Mileage values contain mixed units (kmpl and km/kg), which limits direct comparison.",
    "Potential outliers were retained because some may represent genuine vehicles.",
    "Correlation indicates association and does not establish causation.",
    "The analysis depends on the accuracy of the original dataset."
]

# ============================================================
# 7. SAVE CLEANED DATASET AND TABLES
# ============================================================

df_clean.to_csv(
    "Cars_EDA_Output/Cleaned_Car_Dataset.csv",
    index=False
)

outlier_df.to_csv(
    "Cars_EDA_Output/Outlier_Analysis.csv",
    index=False
)

df_clean.describe().to_csv(
    "Cars_EDA_Output/Descriptive_Statistics.csv"
)

# ============================================================
# 8. GENERATE PDF REPORT
# ============================================================

print("\n========== GENERATING PDF REPORT ==========")

pdf_path = "Cars_EDA_Output/Hasini_Cars_EDA_Report.pdf"

doc = SimpleDocTemplate(
    pdf_path,
    rightMargin=40,
    leftMargin=40,
    topMargin=40,
    bottomMargin=40
)

styles = getSampleStyleSheet()
story = []

title_style = styles['Title']
title_style.alignment = TA_CENTER

story.append(Paragraph(
    "Exploratory Data Analysis on Cars Dataset",
    title_style
))
story.append(Spacer(1, 15))

story.append(Paragraph(
    "<b>Student:</b> Hasini T<br/>"
    "<b>Branch:</b> CSE (Artificial Intelligence and Machine Learning)<br/>"
    "<b>Project:</b> Minor Project - Exploratory Data Analysis",
    styles['Normal']
))
story.append(Spacer(1, 15))

story.append(Paragraph("1. Introduction", styles['Heading1']))
story.append(Paragraph(
    "Exploratory Data Analysis (EDA) is used to understand a dataset "
    "through data inspection, cleaning, statistical summaries and visualizations. "
    "This project analyses a used-car dataset to identify patterns in car brands, "
    "fuel types, selling prices, manufacturing years and vehicle specifications.",
    styles['BodyText']
))

story.append(Spacer(1, 12))
story.append(Paragraph("2. Dataset Overview", styles['Heading1']))
story.append(Paragraph(
    f"The original dataset contains {df.shape[0]} rows and {df.shape[1]} columns. "
    f"After removing duplicate records and handling missing values, "
    f"the cleaned dataset contains {df_clean.shape[0]} rows and "
    f"{df_clean.shape[1]} columns.",
    styles['BodyText']
))

story.append(Spacer(1, 12))
story.append(Paragraph("3. Data Cleaning", styles['Heading1']))
story.append(Paragraph(
    "Duplicate records were removed. Numerical missing values were handled "
    "using median imputation, while missing torque values were labelled as Unknown. "
    "Invalid zero values in mileage and maximum power were treated as missing "
    "and replaced using their respective medians. The cleaned dataset was checked "
    "again for missing values and duplicates.",
    styles['BodyText']
))

story.append(Spacer(1, 12))
story.append(Paragraph("4. Descriptive Statistics", styles['Heading1']))

stats = df_clean.describe().round(2).reset_index()
stats_data = [stats.columns.tolist()] + stats.astype(str).values.tolist()

table = Table(stats_data, repeatRows=1)
table.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.darkblue),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.white),
    ('GRID', (0, 0), (-1, -1), 0.4, colors.grey),
    ('FONTSIZE', (0, 0), (-1, -1), 6),
    ('VALIGN', (0, 0), (-1, -1), 'MIDDLE')
]))
story.append(table)

story.append(PageBreak())

story.append(Paragraph("5. Data Visualizations and Observations", styles['Heading1']))

observations = [
    "Maruti is the most frequently occurring brand in the dataset.",
    "Diesel cars are the most common fuel type, followed by petrol.",
    "Selling prices are positively skewed, with a small number of high-priced cars.",
    "Newer manufacturing years generally show higher selling prices.",
    "Kilometres driven shows a slight negative relationship with selling price.",
    "Automatic cars have a higher average selling price than manual cars.",
    "Maximum power has the strongest positive correlation with selling price among the analysed numerical features.",
    "The boxplots identify potential outliers in numerical features."
]

for i, (path, observation) in enumerate(zip(chart_paths, observations)):
    story.append(Paragraph(
        f"Figure {i+1}: {observation}",
        styles['BodyText']
    ))
    story.append(Spacer(1, 5))
    story.append(Image(path, width=6.3*inch, height=3.4*inch))
    story.append(Spacer(1, 12))

story.append(PageBreak())

story.append(Paragraph("6. Key Findings", styles['Heading1']))

for finding in findings:
    story.append(Paragraph("• " + finding, styles['BodyText']))
    story.append(Spacer(1, 8))

story.append(Paragraph("7. Outlier Analysis", styles['Heading1']))
story.append(Paragraph(
    "The Interquartile Range (IQR) method was used to identify potential "
    "outliers in selling price, kilometres driven, engine size and maximum power. "
    "Potential outliers were retained because extreme values may correspond "
    "to genuine vehicles. They were not automatically removed.",
    styles['BodyText']
))

story.append(Paragraph("8. Limitations", styles['Heading1']))

for item in limitations:
    story.append(Paragraph("• " + item, styles['BodyText']))
    story.append(Spacer(1, 5))

story.append(Paragraph("9. Conclusion", styles['Heading1']))
story.append(Paragraph(
    "The EDA provided an overview of the used-car dataset and revealed "
    "patterns in brands, fuel types, prices and vehicle characteristics. "
    "The analysis also identified data-quality issues and potential outliers. "
    "These findings demonstrate how data cleaning, statistical analysis and "
    "visualization can help understand a real-world dataset.",
    styles['BodyText']
))

doc.build(story)

# ============================================================
# 9. FINAL STATUS
# ============================================================

print("\n====================================")
print("       PROJECT OUTPUTS READY")
print("====================================")

print("1. PDF Report:", pdf_path)
print("2. Cleaned CSV: Cars_EDA_Output/Cleaned_Car_Dataset.csv")
print("3. Outlier Analysis: Cars_EDA_Output/Outlier_Analysis.csv")
print("4. Descriptive Statistics: Cars_EDA_Output/Descriptive_Statistics.csv")
print("5. Charts:", len(chart_paths))

# Download the report and cleaned dataset
from google.colab import files

files.download(pdf_path)
files.download("Cars_EDA_Output/Cleaned_Car_Dataset.csv")

print("\nEDA EXECUTION COMPLETED!")
```
